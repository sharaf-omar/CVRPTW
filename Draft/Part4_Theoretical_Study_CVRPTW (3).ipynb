# 4. Theoretical Study of the Problem

This section classifies the problem we address, the **Capacitated Vehicle Routing Problem with Time Windows (CVRPTW)** [1], in terms of computational complexity. We prove the classification, support it with references, and explain what it implies for the choice of solution methods (without detailing them yet: they are the subject of Section 6).

**Scope: Level A (Advanced).** We study the advanced version of the project, the **CVRPTW**: the CVRP plus one delivery time window per customer, with waiting allowed on site when the vehicle arrives early and late delivery forbidden. The basic CVRP (Level B) is the special case in which every window is $[0,+\infty[$. All the results below are therefore stated and proved for the problem *with* time windows, and §4.5 explains what the windows add to the theoretical picture.

**Notation** (as in Sections 2 and 3). The customers are $N=\{1,\dots,n\}$ and the depot is node $0$. We have at most $K$ identical vehicles of capacity $Q$. Customer $i$ has a demand $q_i$, a service time $s_i$ and a time window $[e_i,l_i]$. Travelling from $i$ to $j$ takes (and costs) $d_{ij}$. The depot has the horizon $[e_0,l_0]$, $q_0=s_0=0$.

A route $r=(0,v_1,\dots,v_m,0)$ is **feasible** if and only if:
1. $\sum_{k=1}^{m} q_{v_k}\le Q$ (capacity);
2. with $v_0=0$ and $b_{v_0}=e_0$, the service start times $b_{v_k}=\max\big(e_{v_k},\,b_{v_{k-1}}+s_{v_{k-1}}+d_{v_{k-1}v_k}\big)$ satisfy $b_{v_k}\le l_{v_k}$ for all $k$ (waiting is allowed, arriving late is forbidden);
3. $b_{v_m}+s_{v_m}+d_{v_m0}\le l_0$ (return to the depot before the horizon ends).

A **solution** is a set of at most $K$ feasible routes in which every customer appears exactly once. Its **cost** is the total distance travelled, which we minimise.

### Summary of the results

| Problem | Complexity class | Where it is justified |
|---|---|---|
| CVRPTW-D (decision version: is there a solution of cost $\le B$?) | **NP-complete** (in the strong sense) | §4.2 and §4.3 |
| CVRPTW (optimisation version: minimise the cost) | **NP-hard** (in the strong sense) | §4.3 |
| Deciding whether a feasible solution exists with a bounded fleet (already for $K=1$, with time windows only) | **NP-complete** | §4.4 (Theorem 2) |
| Minimising the number of vehicles | **NP-hard** | §4.4 |
| Approximation with a guaranteed ratio | impossible in polynomial time, unless P = NP | §4.7 |

## 4.1 Decision version of the problem

The classes P and NP are defined for *decision* problems (yes/no answers). We therefore associate to our optimisation problem the following decision problem.

> **CVRPTW-D.**
> *Input:* an instance $I=\big(n,K,Q,(q_i),(s_i),([e_i,l_i]),(d_{ij})\big)$ and an integer $B$.
> *Question:* does there exist a solution of total cost at most $B$?

The optimisation problem is at least as hard as CVRPTW-D: if we could compute the optimal cost, we could answer the question by comparing it with $B$. Since an optimisation problem is not a yes/no problem, it cannot be "in NP" strictly speaking, and we say that it is **NP-hard** when its decision version is NP-complete. The survey of Toth and Vigo [2] and the chapter of Kallehauge et al. [3] give the standard formulations of this family of problems.

## 4.2 CVRPTW-D belongs to NP

A problem is in NP if a proposed answer ("certificate") can be verified in polynomial time.

**Certificate.** A list of routes $R_1,\dots,R_p$, each one being a sequence of customers. Its size is $O(n\log n)$ bits, which is polynomial in the size of the instance.

**Verification algorithm.**
1. *Every customer exactly once:* scan the routes with a marker array of size $n$. Time $O(n)$.
2. *Number of vehicles:* check that $p\le K$. Time $O(1)$.
3. *Capacity and time windows:* for each route, accumulate the demands and compute the start times $b_{v_k}$ with the recursive formula of the notation above, checking $b_{v_k}\le l_{v_k}$ at every step and the return to the depot at the end. Each customer is processed once, so the total time is $O(n)$.
4. *Cost:* sum the distances of the arcs used and compare with $B$. Time $O(n)$.

The whole verification takes $O(n)$ arithmetic operations on numbers of polynomial bit-length. As is customary in such proofs, we assume constant-time access to an element of a route and to a distance $d_{ij}$ (routes stored as arrays, distances in a matrix) and constant-time arithmetic. The input itself has size $\Theta(n^2)$ because of the distance matrix, so a verification in $O(n)$ is well within polynomial time. Hence **CVRPTW-D $\in$ NP**.

## 4.3 CVRPTW-D is NP-hard: a chain of two reductions

The proof follows a chain of two polynomial-time reductions, starting from a problem known to be NP-complete:

$$\text{Hamiltonian Cycle}\ \le_p\ \text{TSP-D}\ \le_p\ \text{CVRPTW-D}.$$

Hamiltonian Cycle is NP-complete (Karp [4]; Garey and Johnson [5]). Since polynomial reductions compose, the chain shows that CVRPTW-D is at least as hard as Hamiltonian Cycle.

### Step 1: the decision version of the TSP is NP-complete

> **TSP-D.** *Input:* $m+1$ cities $\{0,1,\dots,m\}$, symmetric non-negative integer distances $c_{ij}$ and an integer $B$.
> *Question:* is there a cycle visiting every city exactly once, of total length at most $B$?

This is a classical result [4, 5]. We recall the proof because it is the basis of the whole chain.

*Membership in NP.* Given a cyclic sequence of cities, we check in $O(m)$ time that every city appears exactly once and that the sum of the $m+1$ distances is at most $B$.

*NP-hardness.* Let $G=(V,E)$ be an instance of Hamiltonian Cycle, with $|V|=m+1$ vertices. We build the complete graph on $V$ with the distances $c_{ij}=1$ if $\{i,j\}\in E$ and $c_{ij}=2$ otherwise, and we take $B=|V|$. The construction examines each pair of vertices once, so it takes $O(|V|^2)$ time.
- If $G$ has a Hamiltonian cycle, we use it as the tour: it has $|V|$ edges of length $1$, hence a total length $|V|=B$.
- Conversely, suppose there is a tour of length at most $|V|$. It uses exactly $|V|$ edges, each of length at least $1$, so its length is at least $|V|$. The length is therefore exactly $|V|$, which forces every edge to have length $1$, i.e. to belong to $E$. The tour is a Hamiltonian cycle of $G$.

Note that the bound is $B=|V|$ and not $|V|-1$: a cycle through $|V|$ vertices uses $|V|$ edges. (With the distances $1$ and $2$ the triangle inequality holds, so a closed walk that is allowed to revisit cities can always be shortened to a tour without increasing its length: the variant "at least once" and the variant "exactly once" are equivalent here.)

### Step 2: CVRPTW-D is NP-hard

**Theorem 1.** CVRPTW-D is NP-hard, even when $K=1$, the capacity never binds and the time windows never bind.

**Construction.** From an instance of TSP-D we build, in polynomial time, the following instance of CVRPTW-D, where $c_{\max}=\max_{i,j}c_{ij}$.

| Data of CVRPTW-D | Value chosen | Purpose |
|---|---|---|
| Depot | city $0$ | starting point of the tour |
| Customers | cities $1,\dots,m$ (so $n=m$) | |
| Distances | $d_{ij}=c_{ij}$ | same graph |
| Number of vehicles | $K=1$ | one single route must serve everybody |
| Demands and capacity | $q_i=1$ for all customers, $Q=m$ | the vehicle can carry everything: capacity never binds |
| Service times | $s_i=0$ | |
| Time windows (customers and depot) | $[e_i,l_i]=[0,L]$ with $L=(m+1)\,c_{\max}$ | windows never bind |
| Bound | $B$ (unchanged) | |

**Correctness.** We show that the TSP-D instance is a yes-instance if and only if the CVRPTW-D instance is.

*($\Leftarrow$)* Suppose there is a solution of cost at most $B$. Since $K=1$ and every customer must be served, the solution is a single route that goes from the depot through all $m$ customers exactly once and back to the depot. This is a cycle through all cities, and its length equals the cost, which is at most $B$.

*($\Rightarrow$)* Suppose there is a cycle $(0,v_1,\dots,v_m,0)$ of length at most $B$. We use it as the single route. It is feasible:
- *Capacity:* the load is $\sum q_{v_k}=m=Q$.
- *Time windows:* since $e_i=0$ for all nodes, the vehicle never waits, so the service start time at $v_k$ is $\sum_{j\le k}c_{v_{j-1}v_j}\le (m+1)c_{\max}=L=l_{v_k}$. The same bound holds for the return to the depot.

Its cost is the length of the cycle, which is at most $B$. So the CVRPTW-D instance is a yes-instance.

**Polynomial time and size.** The built instance has $m$ customers and $O(m^2)$ numbers, each one bounded by $L=(m+1)c_{\max}$. It is obtained by copying the distance matrix and computing $L$, which takes $O(m^2)$ time, so its size is polynomial in the size of the TSP-D instance. $\square$

By Step 1 and Step 2, Hamiltonian Cycle reduces to CVRPTW-D in polynomial time. Together with §4.2, **CVRPTW-D is NP-complete**, and the optimisation problem **CVRPTW is NP-hard**.

**Remark 1 (strong NP-hardness).** The distances produced by Step 1 are in $\{1,2\}$ and the bound is $B=m+1$, so all the numbers of the instance built in Step 2 are bounded by a polynomial in $m$ [5]. Therefore CVRPTW is **strongly** NP-hard: there is no pseudo-polynomial algorithm for it unless P = NP.

**Remark 2 (benchmark instances).** The standard benchmarks (Solomon [1], Homberger, Uchoa et al.) use Euclidean distances rounded to a fixed precision. The Euclidean TSP is itself NP-hard (Papadimitriou [6]), and the reduction above keeps the distances unchanged. The hardness therefore holds for the class of instances we will actually solve.

## 4.4 Each family of constraints is hard on its own

The reduction in §4.3 shows that the *routing* part is already hard. Capacity and time windows are hard on their own as well. The time windows, which are specific to Level A, get a proof of their own.

**Capacity.**
- Without time windows (all windows $[0,+\infty[$), CVRPTW becomes the CVRP, which is NP-hard. This was shown by Lenstra and Rinnooy Kan [7].
- If moreover all distances are $0$, the question "do $K$ vehicles suffice?" is exactly the **Bin Packing** problem: customers are items of size $q_i$, vehicles are bins of capacity $Q$. Bin Packing is strongly NP-complete [5]. So *deciding whether a feasible solution exists with $K$ vehicles* is already NP-hard, even with no distances at all.

**Time windows (Level A).** We now show that the windows alone make the problem hard, even with a single vehicle and no cost to minimise.

> **Theorem 2.** Deciding whether a feasible solution exists is NP-complete for CVRPTW, even when $K=1$ and the capacity never binds. (Savelsbergh [8] proved the corresponding result for the TSP with Time Windows.)

*Proof.* Membership in NP is shown as in §4.2. For hardness, we reduce from the **Directed Hamiltonian Path** problem, which is NP-complete [5]: given a directed graph $G=(\{1,\dots,m\},A)$, is there a path visiting every vertex exactly once?

*Construction.* The depot is node $0$ and the customers are the vertices $1,\dots,m$. We take $K=1$, $q_i=1$, $Q=m$ and $s_i=0$. The distances are $d_{0j}=d_{j0}=1$ for all $j$, and for two customers $d_{ij}=1$ if $(i,j)\in A$ and $d_{ij}=2$ otherwise. The customer windows are $[0,m]$ and the depot horizon is $[0,m+1]$. We ask only whether a feasible solution exists.

*Correctness.*
- *($\Leftarrow$)* If $v_1,\dots,v_m$ is a Hamiltonian path, the route $(0,v_1,\dots,v_m,0)$ is feasible. The vehicle never waits, service starts at times $1,2,\dots,m$ (every step has length $1$), all at most $m$, and the vehicle is back at time $m+1$. The load is $m=Q$.
- *($\Rightarrow$)* A feasible solution with $K=1$ is a single route $(0,v_1,\dots,v_m,0)$ through all customers. Every step lasts at least $1$, so $v_k$ is reached at time at least $k$, and each step that is not an arc of $G$ lasts $2$ and adds one more unit of delay. Since $v_m$ must be reached by time $m$, there is no delay at all: every step $(v_{k-1},v_k)$ with $k\ge2$ is an arc of $G$. Hence $v_1,\dots,v_m$ is a Hamiltonian path. $\square$

The difficulty here comes from the windows alone: each customer must be reached early, so the vehicle has no time for anything but the arcs of $G$.

**Role of the fleet size.** The hardness of feasibility relies on the bound $K$ on the number of vehicles. Without it, a feasible solution is easy to build as soon as every customer can be served alone (one route $(0,i,0)$ per customer). What remains hard is to **use few vehicles**: deciding whether $K$ vehicles suffice is NP-hard (Theorem 2 for $K=1$, Bin Packing for the capacity). So *minimising the number of vehicles*, which is the primary objective in the original Solomon benchmarks [1], is NP-hard too.

**Consequence.** For the plain TSP, any permutation of the cities is a valid solution, and the difficulty lies only in the optimisation. For CVRPTW with a bounded fleet, **feasibility itself is hard to achieve**. This changes how algorithms must be designed (see §4.7).

## 4.5 What the time windows change compared with the basic CVRP (Level B)

Choosing the advanced level does not change the complexity class, since the CVRP is already NP-hard and is a special case of CVRPTW. It does change the *nature* of the difficulty, which is what drives the design of the algorithm.

| Aspect | Level B: CVRP | Level A: CVRPTW |
|---|---|---|
| Complexity class | NP-hard [7] | NP-hard (§4.3), it contains the CVRP |
| Sources of hardness | routing and capacity (Bin Packing) | routing, capacity **and time windows** |
| Feasibility of a route | depends only on the *set* of customers (sum of demands) | depends on the **order** of the customers (arrival times) |
| Reversing a route (symmetric distances) | cost and feasibility are unchanged | time flows forward: all arrival times change, and the route may become infeasible |
| Finding a feasible solution with a bounded fleet | Bin Packing (hard) | Bin Packing **and** TSPTW (Theorem 2) |
| Evaluating a modification of a route | constant time with the route loads | propagating the arrival times costs the length of the route, and constant time is possible only with precomputed slack data [12] |
| Extra elements in the ILP | none | time variables $t_i$, big-M constraints, waiting allowed |

**Consequence for our project.** Every move of the algorithm must be checked against the capacity *and* the time windows, and many moves that are harmless in the CVRP become infeasible here. This is why the experimental study will report the feasibility of the solutions, the waiting times and the effect of the window widths, as required for Level A.

## 4.6 Algorithmic analysis: size of the search space

A natural exact algorithm is to enumerate every solution. We count them, ignoring the capacity and time-window constraints (so the count is an upper bound on the number of feasible solutions). A solution is a set of non-empty routes, each route being an ordered list of customers. The number of ways to split $n$ labelled customers into $k$ non-empty ordered lists is the Lah number

$$L(n,k)=\binom{n-1}{k-1}\frac{n!}{k!},$$

so, with at most $K$ vehicles, the search space has size

$$S(n,K)=\sum_{k=1}^{\min(K,n)}\binom{n-1}{k-1}\frac{n!}{k!}\ \ \ge\ n!.$$

Enumeration is therefore $\Omega(n!)$. A smarter exact method for a *single* route is the dynamic programming algorithm of Held and Karp [9], which runs in $O(n^2 2^n)$ time and $O(n\,2^n)$ memory. It is much better than $n!$, but it is still exponential. The table below compares both for a machine that evaluates $10^9$ solutions (or elementary operations) per second.

```python
from math import comb, factorial, log10
import pandas as pd

def n_route_sets(n, K=None):
    # Nombre de solutions possibles, sans contraintes : sum_k Lah(n, k), k <= K
    K = n if K is None else min(K, n)
    return sum(comb(n - 1, k - 1) * (factorial(n) // factorial(k)) for k in range(1, K + 1))

# Vérification de la formule sur des valeurs connues (1, 3, 13, 73, 501)
assert [n_route_sets(n) for n in range(1, 6)] == [1, 3, 13, 73, 501]

SPEED = 1e9                    # solutions évaluées par seconde (hypothèse généreuse)
YEAR = 3.156e7                 # secondes dans une année
rows = []
for n in [5, 10, 15, 20, 25, 30, 50, 100]:
    S = n_route_sets(n)
    seconds = S / SPEED
    rows.append({
        "n (customers)": n,
        "search space S(n)": f"{S:.2e}" if S > 1e6 else str(S),
        "log10 S(n)": round(log10(S), 1),
        "brute force time": (f"{seconds:.1e} s" if seconds < 3600 else
                              f"{seconds / 3600:.1f} h" if seconds < 86400 * 365 else
                              f"{seconds / YEAR:.1e} years"),
        "Held-Karp ops n²·2ⁿ": f"{n * n * 2 ** n:.1e}",
    })
pd.DataFrame(rows).set_index("n (customers)")
```

**Reading the table.** The number of candidate solutions grows super-exponentially. At $n=20$ customers, enumeration would already take about ten thousand years at $10^9$ solutions per second, and at $n=25$ it would take far longer than the age of the universe ($\approx 1.4\times10^{10}$ years). The benchmark instances have 100 customers, and our project targets instances with several thousand customers. Even the exponential dynamic program, which is far more efficient, needs $10^{12}$ operations and tens of billions of memory entries at $n=30$. The constraints remove part of the search space, but not enough to change these orders of magnitude, and checking feasibility is itself hard (§4.4).

## 4.7 Implications for the solution methods

The following consequences guide the choice of method made in Section 6.

**1. No fast exact algorithm is expected.** Since CVRPTW is NP-hard, a polynomial-time exact algorithm would imply P = NP, which is widely believed to be false. Exact approaches remain useful but are limited in size:
- our integer linear program (Section 2) has polynomial size ($O(n^2)$ variables), but solving it requires an exponential search whose lower bounds are weak because of the "big-M" constraints;
- the best exact methods, based on branch-price-and-cut, solve all 56 Solomon instances with 100 customers and 51 of the 60 Gehring and Homberger instances with 200 customers [10]; see also the survey [11]. These algorithms are very sophisticated and computationally heavy, and the results reported in [10] stop at 200 customers, whereas our project targets instances of several thousand customers.

**2. No polynomial-time approximation guarantee in general.** Suppose that a polynomial-time algorithm returned, for some ratio $\rho\ge1$, a feasible solution of cost at most $\rho$ times the optimum whenever a feasible solution exists. In particular it would return a feasible solution whenever one exists, hence decide the feasibility of the TSPTW (one vehicle), which is NP-complete by Theorem 2 [8]. This is impossible unless P = NP. The argument uses a bounded fleet ($K=1$), as in the problem definition. So **no algorithm that runs in polynomial time can guarantee a bounded distance to the optimum** for the general problem.

**3. The quality of a heuristic must be measured experimentally.** Because no guarantee can be proved, the performance of any approximate method is assessed empirically, by comparing its solutions with the best known solutions of reference instances (the *gap*), over many instances and several random seeds. This is exactly the purpose of the experimental study of the final deliverable.

**4. Scalable approximate methods are the realistic choice for large instances.** Heuristics and metaheuristics give up the proof of optimality in exchange for computation times that grow gently with $n$. The choice made among them is detailed and justified in Section 6.

**5. Feasibility must be handled inside the algorithm.** Since finding a feasible solution is already hard (§4.4), a method cannot rely on every move producing a valid solution. It needs a fast feasibility test for each candidate modification of a route (constant-time tests based on the slack of the time windows were introduced by Savelsbergh [12]) and a policy for customers that cannot be inserted without breaking a constraint. The way we do it is described in Section 6.

**6. The hardness is a worst-case statement.** Instances differ a lot in practice: tight time windows prune many combinations, whereas wide windows leave more freedom and make the problem closer to the CVRP. This is why the experimental study will compare families of instances (clustered, random, mixed; short or long scheduling horizon) rather than treat all instances alike.

## 4.8 Conclusion of the theoretical study

| Question | Answer |
|---|---|
| Complexity class of the decision version | NP-complete (strongly) |
| Complexity class of the optimisation version | NP-hard (strongly) |
| Proof technique | membership in NP by an $O(n)$ verification, then reduction from the TSP |
| Source of the difficulty | routing (TSP), capacity (Bin Packing / CVRP) and time windows (TSPTW) |
| What Level A (time windows) adds | feasibility depends on the order of visits and is itself NP-hard with a bounded fleet |
| Consequence for exact methods | exponential time; limited to a few hundred customers |
| Consequence for approximation | no polynomial-time guarantee unless P = NP |
| Consequence for our project | use an approximate method with built-in feasibility handling, evaluated experimentally |
